Importamos Panda

```python
import pandas as pd
```

```python
df = pd.read_csv('data/got.csv')
df.head(10)
```

Agrupamos por asesino(killer)

```python
df_killer = df.groupby('killer')['death_no'].agg('count')
df_killer
```

Quiero mostrar los 5 primeros asesinos.

```python
df_primeros = df_killer.head(5)
df_primeros
```

Grafico dibujado

```python
df_primeros.groupby('killer').sum().plot(kind='pie', y = 'count')
```

Creamos ficheros de asesinos por temporadas

```python
for index in (1,2,3,4,5,6,7,8):
    df_season = df[df['season'] == index]
    df_season.to_csv(f'data/got{index}.csv')
```
